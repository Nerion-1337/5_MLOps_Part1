```python
import pandas as pd

# 1. On charge la base de données propre
df_complet = pd.read_parquet("../data/processed/dataset_train_fusion.parquet") # Adapte le nom/chemin

# 2. On enlève la colonne cible 'TARGET' (car l'API n'est pas censée connaître la réponse !)
X = df_complet.drop(columns=['TARGET']) 

# 3. On prend 50 clients au hasard
X_sample = X.sample(n=50, random_state=42)

# 4. On exporte !
X_sample.to_parquet("../data/X_test_sample.parquet", index=False)
print("✅ Échantillon sauvé !")
```
